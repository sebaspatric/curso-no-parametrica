```
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import levene
```

# Los datos representan los salarios de 3 departamentos: Ventas, Gerente y Administrador.

# 2 grupos

```
link = "https://raw.githubusercontent.com/sebaspatric/curso-no-parametrica/refs/heads/main/2.%20Distribuciones%20y%20varianzas%20de%20grupos/2.5%20Homogeneidad%20de%20varianzas%20en%20Python/homogeneidad2grupos.csv"
```

```
df = pd.read_csv(link)
```

```
#df=pd.read_csv('/content/homogeneidad2grupos.csv')
```

```
df.head()
```

```
group1=df[df.position==1.0]
group2=df[df.position==2.0]
group1.head()
```

```
a = group1.salary
b = group2.salary
stat_mean, p_mean = levene(a, b, center='mean')
stat_med, p_med = levene(a, b, center='median')
print('p-value with mean:',p_mean)
print('p-value with median:',p_med)
```

# 3 grupos

```
link3="https://raw.githubusercontent.com/sebaspatric/curso-no-parametrica/refs/heads/main/2.%20Distribuciones%20y%20varianzas%20de%20grupos/2.5%20Homogeneidad%20de%20varianzas%20en%20Python/homogeneidad3grupos.csv"
```

```
#df=pd.read_csv('/content/homogeneidad3grupos.csv')
```

```
df = pd.read_csv(link3)
```

```
df
```

```
group1=df[df.position==1.0]
group2=df[df.position==2.0]
group3=df[df.position==3.0]
a = group1.salary
b = group2.salary
c = group3.salary
stat_mean, p_mean = levene(a, b, c, center='mean')
stat_med, p_med = levene(a, b, c, center='median')
print('p-value with mean:',p_mean)
print('p-value with median:',p_med)
```
